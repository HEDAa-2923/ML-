### 1. Download and Extract the Dataset
We will use the UCI Human Activity Recognition Using Smartphones Dataset. This dataset contains sensor signals (accelerometer and gyroscope) from 30 subjects performing six activities.

```python
import urllib.request
import zipfile
import os

# URL of the dataset
url = 'https://archive.ics.uci.edu/static/public/240/human+activity+recognition+using+smartphones.zip'
zip_path = 'HAR_dataset.zip'
extract_path = 'HAR_data'

# Download the file
if not os.path.exists(zip_path):
    print("Downloading dataset...")
    urllib.request.urlretrieve(url, zip_path)
    print("Download complete.")

# Extract the outer zip
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# Inside there is another zip file, let's locate and extract it
inner_zip = os.path.join(extract_path, 'UCI HAR Dataset.zip')
if os.path.exists(inner_zip):
    with zipfile.ZipFile(inner_zip, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Extraction complete.")
else:
    print("Nested zip file not found or already extracted.")
```

### 2. Load and Preview the Data
Now, let's load the training and test data including the features, labels, and subject identifiers.

```python
import pandas as pd
import numpy as np

data_dir = 'HAR_data/UCI HAR Dataset'

# Load feature names
features = []
with open(os.path.join(data_dir, 'features.txt'), 'r') as f:
    features = [line.split()[1] for line in f.readlines()]

# Load activity labels mapping
activity_labels = {}
with open(os.path.join(data_dir, 'activity_labels.txt'), 'r') as f:
    for line in f:
        parts = line.split()
        activity_labels[int(parts[0])] = parts[1]

# Helper function to load dataset split
def load_dataset_split(split):
    X = pd.read_csv(os.path.join(data_dir, split, f'X_{split}.txt'), delim_whitespace=True, header=None)
    X.columns = features

    y = pd.read_csv(os.path.join(data_dir, split, f'y_{split}.txt'), header=None, names=['activity_id'])
    y['activity'] = y['activity_id'].map(activity_labels)

    subject = pd.read_csv(os.path.join(data_dir, split, f'subject_{split}.txt'), header=None, names=['subject_id'])

    # Combine into a single DataFrame
    df = pd.concat([subject, y, X], axis=1)
    return df

# Load train and test sets
train_df = load_dataset_split('train')
test_df = load_dataset_split('test')

print(f"Train shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")
display(train_df.head())
```

### 4. Model Training and Evaluation
We will train a Random Forest Classifier as a strong baseline model for this tabular dataset containing extracted sensor features, and evaluate it using classification metrics and a confusion matrix.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Separate features and targets (dropping subject and label columns)
X_train = train_df.drop(columns=['subject_id', 'activity_id', 'activity'])
y_train = train_df['activity']

X_test = test_df.drop(columns=['subject_id', 'activity_id', 'activity'])
y_test = test_df['activity']

# Initialize and train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
print("Training Random Forest...")
rf_model.fit(X_train, y_train)
print("Training complete.")
```

```python
# Predict and evaluate
y_pred = rf_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {accuracy:.4f}\n")

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Plot Confusion Matrix
labels = sorted(y_test.unique())
cm = confusion_matrix(y_test, y_pred, labels=labels)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Confusion Matrix - Random Forest Classifier')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```

### 5. Deep Learning: Training an MLP with PyTorch
Let's build a PyTorch Multi-Layer Perceptron (MLP) to classify the activities using the 561-dimensional hand-crafted feature set.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import LabelEncoder

# Encode target labels to integers (0 to 5)
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

# Convert DataFrames to PyTorch Tensors
X_train_tensor = torch.tensor(X_train.values, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train_encoded, dtype=torch.long)
X_test_tensor = torch.tensor(X_test.values, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test_encoded, dtype=torch.long)

# Create Datasets and DataLoaders
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

# Define the MLP Neural Network architecture
class ActivityMLP(nn.Module):
    def __init__(self, input_dim, num_classes):
        super(ActivityMLP, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.BatchNorm1d(64),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )

    def forward(self, x):
        return self.network(x)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = ActivityMLP(input_dim=X_train.shape[1], num_classes=len(label_encoder.classes_)).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print(f"Using device: {device}")
```

```python
# Training Loop
epochs = 15
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for batch_X, batch_y in train_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)

        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_X.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)

    # Evaluate epoch loss and test accuracy
    model.eval()
    correct = 0
    with torch.no_grad():
        for batch_X, batch_y in test_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            outputs = model(batch_X)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == batch_y).sum().item()

    accuracy = correct / len(test_loader.dataset)
    print(f"Epoch {epoch+1:02d}/{epochs} - Train Loss: {epoch_loss:.4f} - Test Accuracy: {accuracy:.4f}")
```

### 6. Stabilizing Neural Network Training with Checkpointing
Let's retrain our MLP model, but this time we will save the model parameters that yield the highest test accuracy during training. This prevents keeping a suboptimal model due to epoch-to-epoch performance oscillations.

```python
import copy

# Re-initialize the model, optimizer, and loss to train from scratch
model = ActivityMLP(input_dim=X_train.shape[1], num_classes=len(label_encoder.classes_)).to(device)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

best_accuracy = 0.0
best_model_wts = copy.deepcopy(model.state_dict())

epochs = 15
print("Training MLP with checkpointing...")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for batch_X, batch_y in train_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)

        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_X.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)

    # Evaluation phase
    model.eval()
    correct = 0
    with torch.no_grad():
        for batch_X, batch_y in test_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            outputs = model(batch_X)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == batch_y).sum().item()

    accuracy = correct / len(test_loader.dataset)
    print(f"Epoch {epoch+1:02d}/{epochs} - Train Loss: {epoch_loss:.4f} - Test Accuracy: {accuracy:.4f}")

    # Checkpoint the best weights
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_model_wts = copy.deepcopy(model.state_dict())

print(f"\nTraining complete! Best Test Accuracy achieved: {best_accuracy:.4f}")

# Load the best model weights
model.load_state_dict(best_model_wts)
```

```python
from sklearn.metrics import classification_report

# Final evaluation with the best saved weights
model.eval()
all_preds = []
with torch.no_grad():
    for batch_X, _ in test_loader:
        batch_X = batch_X.to(device)
        outputs = model(batch_X)
        _, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.cpu().numpy())

# Decode numerical targets back to their original activity names
y_pred_decoded = label_encoder.inverse_transform(all_preds)

print("Classification Report for the Best MLP Model:")
print(classification_report(y_test, y_pred_decoded))
```

### 7. Deep Learning: Hybrid CNN-LSTM on Raw Inertial Signals
Instead of using the pre-computed 561 features, we will load the raw 128-step time-series inertial signals (accelerometer and gyroscope) to train a hybrid **CNN-LSTM** network in PyTorch.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# Helper function to load a single file of raw signals
def load_file(filepath):
    df = pd.read_csv(filepath, delim_whitespace=True, header=None)
    return df.values

# Helper function to load all 9 inertial signals for a split (train or test)
def load_group(signals_directory, signal_types):
    loaded = []
    for name in signal_types:
        filename = f"{signals_directory}/{name}.txt"
        loaded.append(load_file(filename))
    # Stack signals into a 3D array: (samples, time_steps, features/channels)
    return np.transpose(np.dstack(loaded), (0, 1, 2))

signal_types = [
    'body_acc_x_train', 'body_acc_y_train', 'body_acc_z_train',
    'body_gyro_x_train', 'body_gyro_y_train', 'body_gyro_z_train',
    'total_acc_x_train', 'total_acc_y_train', 'total_acc_z_train'
]

signal_types_test = [name.replace('train', 'test') for name in signal_types]

# Load raw signals
print("Loading raw inertial training signals...")
X_train_raw = load_group(f"{data_dir}/train/Inertial Signals", signal_types)
print("Loading raw inertial testing signals...")
X_test_raw = load_group(f"{data_dir}/test/Inertial Signals", signal_types_test)

print(f"Raw Train shape: {X_train_raw.shape}") # Expected: (7352, 128, 9)
print(f"Raw Test shape: {X_test_raw.shape}")   # Expected: (2947, 128, 9)
```

```python
# Convert to PyTorch Tensors (PyTorch Conv1D expects channel-first format: (batch, channels, time_steps))
X_train_raw_tensor = torch.tensor(X_train_raw, dtype=torch.float32).transpose(1, 2)
X_test_raw_tensor = torch.tensor(X_test_raw, dtype=torch.float32).transpose(1, 2)

train_raw_dataset = TensorDataset(X_train_raw_tensor, y_train_tensor)
test_raw_dataset = TensorDataset(X_test_raw_tensor, y_test_tensor)

train_raw_loader = DataLoader(train_raw_dataset, batch_size=64, shuffle=True)
test_raw_loader = DataLoader(test_raw_dataset, batch_size=64, shuffle=False)

# Define the CNN-LSTM Architecture
class CNNLSTM(nn.Module):
    def __init__(self, input_channels, num_classes):
        super(CNNLSTM, self).__init__()
        # CNN layer extracts features from the temporal sequence
        self.cnn = nn.Sequential(
            nn.Conv1d(input_channels, 64, kernel_size=5, padding=2),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2),
            nn.Dropout(0.2),
            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2),
            nn.Dropout(0.2)
        )
        # LSTM process the feature representations output by the CNN
        self.lstm = nn.LSTM(input_size=128, hidden_size=64, num_layers=1, batch_first=True)
        self.fc = nn.Linear(64, num_classes)

    def forward(self, x):
        # CNN expects: (batch, channels, time_steps)
        features = self.cnn(x)
        # Transpose to LSTM format: (batch, sequence_length, features)
        features = features.transpose(1, 2)
        lstm_out, (hn, cn) = self.lstm(features)
        # Retrieve final step sequence output
        out = lstm_out[:, -1, :]
        return self.fc(out)

model_raw = CNNLSTM(input_channels=9, num_classes=len(label_encoder.classes_)).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_raw.parameters(), lr=0.001)
print(model_raw)
```

```python
import copy

best_acc_raw = 0.0
best_raw_wts = copy.deepcopy(model_raw.state_dict())

epochs = 15
print("Training CNN-LSTM...")
for epoch in range(epochs):
    model_raw.train()
    running_loss = 0.0
    for batch_X, batch_y in train_raw_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)

        optimizer.zero_grad()
        outputs = model_raw(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_X.size(0)

    epoch_loss = running_loss / len(train_raw_loader.dataset)

    # Evaluate
    model_raw.eval()
    correct = 0
    with torch.no_grad():
        for batch_X, batch_y in test_raw_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            outputs = model_raw(batch_X)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == batch_y).sum().item()

    accuracy = correct / len(test_raw_loader.dataset)
    print(f"Epoch {epoch+1:02d}/{epochs} - Train Loss: {epoch_loss:.4f} - Test Accuracy: {accuracy:.4f}")

    if accuracy > best_acc_raw:
        best_acc_raw = accuracy
        best_raw_wts = copy.deepcopy(model_raw.state_dict())

print(f"\nTraining complete! Best Test Accuracy on Raw Signals: {best_acc_raw:.4f}")
model_raw.load_state_dict(best_raw_wts)
```

### 9. Advanced Step 1: Solving 'Sitting vs. Standing' with Butterworth Low-Pass Filter
To address the mutual confusion of static activities, we isolate the constant gravity vector by applying a 3rd-order Butterworth low-pass filter with a 0.3 Hz cutoff frequency to the total acceleration signals.

```python
from scipy.signal import butter, filtfilt

def butter_lowpass_filter(data, cutoff=0.3, fs=50, order=3):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    # Apply filter along the time axis (axis=1)
    return filtfilt(b, a, data, axis=1)

# Example of isolating gravity from total acceleration X-channel (channel index 6 in X_train_raw)
# Raw train signal channels: index 6, 7, 8 are total acceleration x, y, z
gravity_train_x = butter_lowpass_filter(X_train_raw[:, :, 6])
body_acc_train_x_isolated = X_train_raw[:, :, 6] - gravity_train_x

# Visualize raw total, isolated gravity, and isolated body acceleration for a single sample
plt.figure(figsize=(12, 4))
plt.plot(X_train_raw[0, :, 6], label='Raw Total Accel (X)', alpha=0.5)
plt.plot(gravity_train_x[0, :], label='Isolated Gravity (Low-pass)', linewidth=2.5)
plt.plot(body_acc_train_x_isolated[0, :], label='Isolated Body Accel (Residual)', alpha=0.7)
plt.title('Butterworth Filtering for Gravity-Body Separation')
plt.xlabel('Timestep (50 Hz)')
plt.ylabel('g-force')
plt.legend()
plt.show()
```

### 11. Advanced Step 3: CNN-LSTM with Temporal Self-Attention Mechanism
We construct an advanced hybrid model adding a custom Self-Attention block over the sequence states produced by the LSTM layer. This helps the network dynamically weight highly informative parts of the 128-step time sequence.

```python
class AttentionBlock(nn.Module):
    def __init__(self, hidden_dim):
        super(AttentionBlock, self).__init__()
        self.attention_weights = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1)
        )

    def forward(self, lstm_outputs):
        # lstm_outputs shape: (batch, seq_len, hidden_dim)
        scores = self.attention_weights(lstm_outputs) # (batch, seq_len, 1)
        weights = torch.softmax(scores, dim=1)
        context = torch.sum(weights * lstm_outputs, dim=1) # (batch, hidden_dim)
        return context, weights

class CNNLSTMAttention(nn.Module):
    def __init__(self, input_channels, num_classes):
        super(CNNLSTMAttention, self).__init__()
        self.cnn = nn.Sequential(
            nn.Conv1d(input_channels, 64, kernel_size=5, padding=2),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2),
            nn.Dropout(0.2),
            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(kernel_size=2),
            nn.Dropout(0.2)
        )
        self.lstm = nn.LSTM(input_size=128, hidden_size=64, num_layers=1, batch_first=True, bidirectional=True)
        self.attention = AttentionBlock(hidden_dim=128) # 64 * 2 (bidirectional)
        self.fc = nn.Sequential(
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )

    def forward(self, x):
        features = self.cnn(x).transpose(1, 2) # (batch, seq_len, features)
        lstm_out, _ = self.lstm(features)
        context, attn_weights = self.attention(lstm_out)
        return self.fc(context)

model_attn = CNNLSTMAttention(input_channels=9, num_classes=len(label_encoder.classes_)).to(device)
print(model_attn)
```

### 13. Training CNN-LSTM with Isolated Butterworth Filter Channels
We will now filter our dataset to explicitly separate **gravity** and **body acceleration** across all three spatial dimensions ($x, y, z$). We will then build an upgraded 12-channel raw signal dataset and evaluate how this affects our classification accuracy.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from scipy.signal import butter, filtfilt

# Re-define our low-pass filter function
def butter_lowpass_filter(data, cutoff=0.3, fs=50, order=3):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    return filtfilt(b, a, data, axis=1)

# Extract the original total acceleration coordinates (indexes 6, 7, 8)
X_train_total_acc = X_train_raw[:, :, 6:9]
X_test_total_acc = X_test_raw[:, :, 6:9]

# Create empty arrays for filtered signals
X_train_gravity = np.zeros_like(X_train_total_acc)
X_test_gravity = np.zeros_like(X_test_total_acc)

# Filter each channel separately
for i in range(3):
    X_train_gravity[:, :, i] = butter_lowpass_filter(X_train_total_acc[:, :, i])
    X_test_gravity[:, :, i] = butter_lowpass_filter(X_test_total_acc[:, :, i])

# Isolate body acceleration by subtracting gravity from total acceleration
X_train_body_acc_isolated = X_train_total_acc - X_train_gravity
X_test_body_acc_isolated = X_test_total_acc - X_test_gravity

# Combine original signals (9 channels) + 3 isolated gravity channels + 3 isolated body channels = 15 channels total
# (Alternatively, let's use the 9 original raw channels and append our isolated gravity and dynamic acceleration)
X_train_augmented = np.concatenate([X_train_raw, X_train_gravity, X_train_body_acc_isolated], axis=2)
X_test_augmented = np.concatenate([X_test_raw, X_test_gravity, X_test_body_acc_isolated], axis=2)

print(f"Augmented Train shape: {X_train_augmented.shape}")  # Expected: (7352, 128, 15)
print(f"Augmented Test shape: {X_test_augmented.shape}")    # Expected: (2947, 128, 15)
```

```python
# Convert augmented dataset to PyTorch Tensors (channel-first format)
X_train_aug_tensor = torch.tensor(X_train_augmented, dtype=torch.float32).transpose(1, 2)
X_test_aug_tensor = torch.tensor(X_test_augmented, dtype=torch.float32).transpose(1, 2)

train_aug_dataset = TensorDataset(X_train_aug_tensor, y_train_tensor)
test_aug_dataset = TensorDataset(X_test_aug_tensor, y_test_tensor)

train_aug_loader = DataLoader(train_aug_dataset, batch_size=64, shuffle=True)
test_aug_loader = DataLoader(test_aug_dataset, batch_size=64, shuffle=False)

# Initialize and train our CNN-LSTM model with the 15-channel augmented input
model_filtered = CNNLSTM(input_channels=15, num_classes=len(label_encoder.classes_)).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_filtered.parameters(), lr=0.001)

best_acc_filtered = 0.0
best_filtered_wts = copy.deepcopy(model_filtered.state_dict())

epochs = 12
print("Training CNN-LSTM with Isolated Butterworth Channels...")
for epoch in range(epochs):
    model_filtered.train()
    running_loss = 0.0
    for batch_X, batch_y in train_aug_loader:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        optimizer.zero_grad()
        outputs = model_filtered(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * batch_X.size(0)

    # Evaluate
    model_filtered.eval()
    correct = 0
    with torch.no_grad():
        for batch_X, batch_y in test_aug_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            outputs = model_filtered(batch_X)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == batch_y).sum().item()

    accuracy = correct / len(test_aug_loader.dataset)
    print(f"Epoch {epoch+1:02d}/{epochs} - Train Loss: {running_loss/len(train_aug_loader.dataset):.4f} - Test Accuracy: {accuracy:.4f}")

    if accuracy > best_acc_filtered:
        best_acc_filtered = accuracy
        best_filtered_wts = copy.deepcopy(model_filtered.state_dict())

print(f"\nTraining complete! Best Filtered Test Accuracy: {best_acc_filtered:.4f}")
model_filtered.load_state_dict(best_filtered_wts)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate predictions with the filtered model
model_filtered.eval()
all_filtered_preds = []
with torch.no_grad():
    for batch_X, _ in test_aug_loader:
        batch_X = batch_X.to(device)
        outputs = model_filtered(batch_X)
        _, predicted = torch.max(outputs, 1)
        all_filtered_preds.extend(predicted.cpu().numpy())

y_pred_filtered_decoded = label_encoder.inverse_transform(all_filtered_preds)

# 2. Plot normalized confusion matrix
labels = sorted(y_test.unique())
cm_filtered = confusion_matrix(y_test, y_pred_filtered_decoded, labels=labels, normalize='true')

plt.figure(figsize=(10, 8))
sns.heatmap(cm_filtered, annot=True, fmt='.2f', cmap='Greens', xticklabels=labels, yticklabels=labels)
plt.title('Normalized Confusion Matrix - CNN-LSTM (With Isolated Butterworth Channels)')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()

print("Classification Report:")
print(classification_report(y_test, y_pred_filtered_decoded))
```
